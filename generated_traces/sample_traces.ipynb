```python
import numpy as np
import pandas as pd
from IPython.display import display

# -----------------------------
# Load NPZ
# -----------------------------
path = "context_4096_gen.npz"   # change if needed
data = np.load(path)

keys = data.files   # all arrays in the NPZ

# -----------------------------
# FILE OVERVIEW
# -----------------------------
print("\n=== FILE OVERVIEW ===")
for k in keys:
    arr = data[k]
    print(f"{k:>5} shape: {arr.shape} {arr.dtype}")

# -----------------------------
# VALUE RANGES
# -----------------------------
print("\n=== VALUE RANGES ===")
for k in keys:
    arr = data[k]
    print(f"{k:>5} min/max: {arr.min()} {arr.max()}")

# -----------------------------
# SHOW ONE SEQUENCE (ALL COLUMNS)
# -----------------------------
seq_id  = 0      # which sequence
n_steps = 20     # positions to show

print(f"\n=== SEQUENCE {seq_id} (positions 0-{n_steps-1}) ===")

rows = []
for pos in range(n_steps):
    row = {"pos": pos}
    for k in keys:
        row[k] = int(data[k][seq_id, pos])
    rows.append(row)

df = pd.DataFrame(rows)
display(df)
```
